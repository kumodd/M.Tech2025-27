```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# Function to plot PMF (probability mass function) and CDF (cumulative distribution function)
def plot_dicrete_dist(dist_name, x_values, pmf_values,cdf_values):
   plt.figure(figsize=(10,4))

   # Plot PMF
   plt.subplot(1,2,1)
   plt.stem(x_values,pmf_values)
   plt.title(f"PMF plot of {dist_name} distribution")
   plt.xlabel("k")
   plt.ylabel("P(X=k)")
   plt.grid(True)

   # Plot CDF
   plt.subplot(1,2,2)
   # plt.subplot(number of row, number col, which number of plot yor are ploting out of number of row*number of col)
   plt.step(x_values,cdf_values)
   plt.title(f"CDF plot of {dist_name} distribution")
   plt.xlabel("k")
   plt.ylabel("P(X<=k)")
   plt.grid(True)
```

```python
# Binomial distribution without using scipy
from math import comb

def binomial_dist(n,p):
   x=np.arange(0,n+1)
   pmf=[comb(n,k)*(p**k)*((1-p)**(n-k)) for k in x]
   cdf=np.cumsum(pmf)
   return x, pmf, cdf
```

```python
n=int(input("Enter the number of trials "))
p=0.9
x_binom, pmf_binom,cdf_binom= binomial_dist(n,p)
dist_name="Binomial"
plot_dicrete_dist(dist_name, x_binom, pmf_binom,cdf_binom)
```
