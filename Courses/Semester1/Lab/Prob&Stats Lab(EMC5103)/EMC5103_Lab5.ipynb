```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# Function to plot PMF (probability mass function) and CDF (cumulative distribution function)
def plot_dicrete_dist(dist_name, x_values, pmf_values,cdf_values):
   plt.figure(figsize=(10,4))

   # Plot PMF
   plt.subplot(1,2,1)
   plt.stem(x_values,pmf_values)
   plt.title(f"PMF plot of {dist_name} distribution")
   plt.xlabel("k")
   plt.ylabel("P(X=k)")
   plt.grid(True)

   # Plot CDF
   plt.subplot(1,2,2)
   # plt.subplot(number of row, number col, which number of plot yor are ploting out of number of row*number of col)
   plt.step(x_values,cdf_values)
   plt.title(f"CDF plot of {dist_name} distribution")
   plt.xlabel("k")
   plt.ylabel("P(X<=k)")
   plt.grid(True)
```

```python
# Binomial distribution without using scipy
from math import comb

def binomial_dist(n,p):
   x=np.arange(0,n+1)
   pmf=[comb(n,k)*(p**k)*((1-p)**(n-k)) for k in x]
   cdf=np.cumsum(pmf)
   return x, pmf, cdf
```

```python
n=int(input("Enter the number of trials "))
p=0.9
x_binom, pmf_binom,cdf_binom= binomial_dist(n,p)
dist_name="Binomial"
plot_dicrete_dist(dist_name, x_binom, pmf_binom,cdf_binom)
```

```python
# Binomial distribution without using scipy
from math import comb
import numpy as np

def binomial_exp(n,p):
   x=np.arange(0,n+1)
   pmf=[comb(n,k)*(p**k)*((1-p)**(n-k)) for k in x]

   exp_x=np.sum([k*pmf[k] for k in x])
   exp_xx=np.sum([(k**2)*pmf[k] for k in x])
   var_b=exp_xx-(exp_x)**2

   return exp_x,var_b
```

```python
n=10
p=0.5
mean_b,var_b=binomial_exp(n,p)

print(f"Mean of the binomial distribution is {mean_b}")
print(f"Variance of the binomial distribution is {var_b}")
```

```python
from math import comb
import numpy as np

def binomial_gamma(n,p):
   x=np.arange(0,n+1)
   pmf=[comb(n,k)*(p**k)*((1-p)**(n-k)) for k in x]

   exp_x=np.sum([k*pmf[k] for k in x])
   mu_2=np.sum([((k-exp_x)**2)*pmf[k] for k in x])
   mu_3=np.sum([((k-exp_x)**3)*pmf[k] for k in x])
   mu_4=np.sum([((k-exp_x)**4)*pmf[k] for k in x])
   gamma_1=mu_3/(mu_2**(3/2))
   gamma_2=mu_4/(mu_2**2)-3
   return gamma_1,gamma_2
```

```python
n=10
p=0.2
gamma1,gamma2=binomial_gamma(n,p)
print(f"The skewness coefficient is {gamma1}")
print(f"The kurtosis coefficient is {gamma2}")
```
