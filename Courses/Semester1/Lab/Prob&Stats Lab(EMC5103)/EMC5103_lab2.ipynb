```python
# Swapping of two number
a=int(input("Enter the first number "))
b=int(input("Enter the second number "))

print(f"Before swapping: a= {a} and b = {b}")

temp=a
a=b
b=temp

print(f"After swapping: a= {a} and b = {b}")
```

```python
# Fibonacci sequence
n=int(input("Enter how many terms you want"))

fib = [0, 1]

for i in range(2,n):
    fib.append(fib[i-1]+fib[i-2])

print(f"The Fibonacci sequence: {fib}")
```

```python
# Factorial of given number
n=int(input("Enter your choice "))

factorial = 1

if n < 0:
   print(f"Factorial of given number {n} is not possible")
elif n==0 or n==1:
   print(f"The factorial of given number {n} is {1}")
else:
   for i in range(1,n+1):
        factorial = i*factorial
   print(f"The factorial of given number {n} is {factorial}")
```

```python
# Numpy library
import numpy as np

# Create vectors
v1=np.array([1,2,3])
v2=np.array([4,5,6])

print(f"vector 1 is: {v1}")
print(f"vector 2 is: {v2}")

print(f"The addition of two vectors is: {v1+v2}")
print(f"Element-wise multiplication of two vectors is: {v1*v2}")
print(f"Element-wise division of two vectors is: {v1/v2}")
print(f"Add any scalar in vector: {3+v1}")
print(f"Multiplication of any scalar in vector: {3*v1}")

print(f"Dot product of any two vectors is: {np.dot(v1,v2)}")
print(f"Cross product of two vectors is {np.cross(v1,v2)}")
```

```python
v=np.array([1,2,3])
print(f"Minima of given array is {np.min(v)}")
print(f"Maxima of given array is {np.max(v)}")
print(f"Mean of given array is {np.mean(v)}")
print(f"Median of given array is {np.median(v)}")
print(f"variance of given array is {np.var(v)}")
```

```python
# Define a list from the user
A=[]

n=int(input("Enter the size of list"))

for i in range(0,n):
  val=int(input(f"Enter the {i}^th element of the list"))
  A.append(val)
print(A)

min_val=A[0]
for num in A:
    if num<min_val:
      min_val=num

print(f"The minima of given list is {min_val}")
```
